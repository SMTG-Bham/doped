# 1. Parse Host structure

```python
from doped.chemical_potentials import get_entries
mgo_entries = get_entries("MgO")
mgo_entry = sorted(mgo_entries, key=lambda x: x.energy_per_atom)[0]  # take lowest energy entry
```

```python
from doped.utils.symmetry import get_primitive_structure
prim_struc = get_primitive_structure(mgo_entry.structure)  # get a clean primitive structure
```

```python
from doped.utils.parsing import qe_convergence_setup_from_structure, qe_relax_setup_from_structure
qe_convergence_setup = qe_convergence_setup_from_structure(prim_struc,
      output_dir="MgO_convergence",
      kpoint_density_range=(20, 200, 20),  # Å^-3, range-style
      ecut_range=(30, 90, 10),             # Ry, inclusive
      ecut_sweep_kpoint_density=100,       # k-density held fixed during ecut sweep
      is_metal=False,
)
```
